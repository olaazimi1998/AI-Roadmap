```python
from sklearn.tree import DecisionTreeClassifier

X = [
    [10],
    [12],
    [14],
    [30],
    [32],
    [35]
]

y = [
    0,
    0,
    0,
    1,
    1,
    1
]

model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=2,
    random_state=42
)

model.fit(X, y)

print(model.predict([[11]]))
print(model.predict([[33]]))
```

Split:English: Dividing a node into two or more sub-nodes based on a feature condition.   فارسی: تقسیم کردن یک گره به چند زیرگره بر اساس یک شرط روی ویژگی‌ها.   Pure vs mixed groups:English: A pure group contains samples of only one class, whereas a mixed group contains samples from multiple classes.   فارسی: گروه خالص فقط شامل یک کلاس است، در حالی که گروه مخلوط شامل داده‌هایی از چند کلاس مختلف می‌باشد.   Gini Impurity:English: A metric measuring the frequency with which a randomly chosen element would be incorrectly labeled.   فارسی: معیاری برای سنجش میزان ناخالصی یا ترکیب نادرست داده‌ها در یک گره.   criterion="gini":English: The parameter in Scikit-Learn that instructs the Decision Tree to use Gini Impurity for splitting nodes.   فارسی: پارامتری در سکایت-لرن که به درخت تصمیم می‌گوید از معیار ناخالصی جینی برای تقسیم‌بندی استفاده کند.   Lower Gini = purer group:English: A lower Gini value indicates less mixing and higher purity in the group (0.0 means perfectly pure).   فارسی: هرچه مقدار جینی کمتر باشد، ناخالصی کمتر و گروه خالص‌تر است (مقدار ۰.۰ یعنی خلوص کامل).   fit():English: The method used to train the machine learning model on the input data and targets.   فارسی: متدی که برای آموزش دادن مدل با استفاده از داده‌های ورودی استفاده می‌شود.   predict():English: The method used to generate predictions for new or unseen data using the trained model.   فارسی: متدی که برای پیش‌بینی خروجی روی داده‌های جدید به کار می‌رود.   max_depth:English: A hyperparameter that limits the maximum depth/layers of the Decision Tree to prevent it from growing too large.   فارسی: هایپرپارامتری که حداکثر عمق یا تعداد لایه‌های درخت را برای جلوگیری از بزرگ شدن بیش از حد محدود می‌کند.   Overfitting:English: A scenario where the model learns the training data and noise too well, causing poor generalization to new data.   فارسی: حالتی که مدل جزئیات و نویزهای داده‌های آموزشی را حفظ کرده و روی داده‌های جدید عملکرد ضعیفی نشان می‌دهد.
