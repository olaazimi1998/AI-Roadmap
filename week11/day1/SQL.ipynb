### SQL = Structured Query Language

SQL is a language used to communicate with relational databases.

For example, imagine you have an AI application that stores users:



```python
| id | name | age | city      |
| -: | ---- | --: | --------- |
|  1 | Ali  |  25 | Dubai     |
|  2 | Sara |  22 | Abu Dhabi |
|  3 | Omar |  30 | Dubai     |
```

```python
%%%sql
SELECT  *
FROM users
WHERE city = 'Dubai';
```
