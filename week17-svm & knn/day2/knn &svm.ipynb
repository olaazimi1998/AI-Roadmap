. What is KNN?
KNN = K-Nearest Neighbors

The idea is very simple:

Look at the closest examples and let them vote.

Example
Suppose we want to classify this new point:

       🔵  🔵

             ❓

       🔴  🔴
If the nearest points to ❓ are:

🔵 🔵 🔴
There are more 🔵 points.

So KNN says:

Prediction = 🔵
2. What is k?
k means:

How many neighbors should we look at?

For example:

k = 3
KNN looks at the 3 closest points.

If:

2 blue
1 red
then:

prediction = blue
If:

1 blue
2 red
then:

prediction = red
Different k
k = 1
Look at only the closest point.

k = 3
Look at the 3 closest points.

k = 5
Look at the 5 closest points.


from sklearn.neighbors import KNeighborsClassifier

X = [
    [1, 1],
    [2, 1],
    [1, 2],
    [6, 6],
    [7, 6],
    [6, 7]
]

y = [0, 0, 0, 1, 1, 1]


model = KNeighborsClassifier(n_neighbors=3)

model.fit(X, y)

prediction = model.predict([
    [2, 2],
    [7, 7]
])

print(prediction)

5. Distance
KNN needs to know:

Which points are close?

The most common idea is Euclidean distance.

For two points:

A = (1, 2)
B = (4, 6)
the distance is:

√((4-1)² + (6-2)²)
You don't need to memorize the mathematics yet.

Just remember:

KNN uses distance to find the nearest neighbors.

Why are we testing different k values?
Because k affects the prediction.

Think about it like this:

k = 1
↓
Only 1 neighbor

k = 3
↓
3 neighbors vote

k = 5
↓
5 neighbors vote
There isn't one magical k that always works.

```python
from sklearn.neighbors import KNeighborsClassifier

X = [
    [1, 1],
    [2, 1],
    [1, 2],
    [6, 6],
    [7, 6],
    [6, 7]
]

y = [0, 0, 0, 1, 1, 1]

model = KNeighborsClassifier(n_neighbors=1)
model.fit(X, y)

prediction = model.predict([
    [2, 2],
    [7, 7]
])

print(prediction)
```

```python
from sklearn.neighbors import KNeighborsClassifier

X = [
    [1, 1],
    [2, 1],
    [1, 2],
    [2, 2],

    [7, 7],
    [8, 7],
    [7, 8],
    [8, 8]
]

y = [
    0, 0, 0, 0,
    1, 1, 1, 1
]


# k = 1
model1 = KNeighborsClassifier(n_neighbors=1)
model1.fit(X, y)

print("k = 1:")
print(model1.predict([[3, 3], [7, 6]]))


# k = 3
model3 = KNeighborsClassifier(n_neighbors=3)
model3.fit(X, y)

print("k = 3:")
print(model3.predict([[3, 3], [7, 6]]))


# k = 5
model5 = KNeighborsClassifier(n_neighbors=5)
model5.fit(X, y)

print("k = 5:")
print(model5.predict([[3, 3], [7, 6]]))
```

```python
from sklearn import svm

X = [
    [1, 1],
    [2, 1],
    [1, 2],
    [6, 6],
    [7, 6],
    [6, 7]
]

y = [0, 0, 0, 1, 1, 1]


# Linear SVM
linear_model = svm.SVC(kernel="linear")

linear_model.fit(X, y)

print("Linear prediction:")
print(linear_model.predict([[2, 2], [7, 7]]))


# RBF SVM
rbf_model = svm.SVC(kernel="rbf")

rbf_model.fit(X, y)

print("RBF prediction:")
print(rbf_model.predict([[2, 2], [7, 7]]))
```
