```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
df = pd.read_csv("C:\\Users\\olaaz\\Documents\\AI Roadmap\\First-project\\day1\\products.csv")
```

```python
df.head(5)
```

```python
sns.barplot(
    x="name",
    y="sell_price",
    hue="brand",
    data=df
)
plt.xticks(rotation=45, ha="right")
plt.title("Sell Price by Brand")
plt.tight_layout()
plt.show()
```

```python
sns.barplot(
    x="name",
    y="buy_price",
    hue="brand",
    data=df
)
plt.xticks(rotation=45, ha="right")
plt.title("buy price by Brand")
plt.tight_layout()
plt.show()
```

```python
sns.barplot(
    x="name",
    y="profit",
    hue="brand",
    data=df
)
sns.scatterplot(
    x="name",
    y="profit",
    hue="brand",
    data=df
)
plt.legend(loc="upper right")
plt.xticks(rotation=45, ha="right")
plt.title("profit by Brand")
plt.tight_layout()
plt.show()
```

```python
sns.barplot(
    x="name",
    y="profit",
    hue="brand",
    data=df
)
sns.scatterplot(
    x="name",
    y="profit",
    hue="brand",
    data=df
)
sns.set(style="darkgrid")
plt.legend(loc="upper right")
plt.xticks(rotation=45, ha="right")
plt.title("profit by Brand")
plt.tight_layout()
plt.show()
```
