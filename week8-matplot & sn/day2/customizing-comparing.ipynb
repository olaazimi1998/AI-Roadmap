```python
months = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug"
]

product_a = [
    120, 150, 140, 180,
    210, 250, 270, 300
]

product_b = [
    100, 130, 160, 170,
    190, 220, 230, 260
]
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(5, 5))
plt.plot(months, product_a, marker="o", label="product_a")
plt.plot(months, product_b, marker="o", label="product_b")
plt.title("Sales month")
plt.xlabel("months")
plt.ylabel("sales")
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1) Insight: Which month had the highest sales for each product?
best_a_month = months[product_a.index(max(product_a))]
best_b_month = months[product_b.index(max(product_b))]
best_a_value = max(product_a)
best_b_value = max(product_b)

print(f"Insight 1: Product A peak sales were in {best_a_month} with {best_a_value} units.")
print(f"Insight 1: Product B peak sales were in {best_b_month} with {best_b_value} units.")

# 2) Insight: Average monthly sales and total difference
avg_a = sum(product_a) / len(product_a)
avg_b = sum(product_b) / len(product_b)
total_a = sum(product_a)
total_b = sum(product_b)

print(f"Insight 2: Product A average monthly sales = {avg_a:.1f}, total = {total_a}.")
print(f"Insight 2: Product B average monthly sales = {avg_b:.1f}, total = {total_b}.")
print(f"Insight 2: Product A outsold Product B by {total_a - total_b} units overall.")

# 3) Insight: Product share and monthly comparison
product_total = [total_a, total_b]
product_names = ["Product A", "Product B"]
share_labels = [f"{name} ({value})" for name, value in zip(product_names, product_total)]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Monthly comparison
axes[0].plot(months, product_a, marker="o", label="Product A", linewidth=2)
axes[0].plot(months, product_b, marker="o", label="Product B", linewidth=2)
axes[0].set_title("Monthly sales comparison")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Sales")
axes[0].legend()
axes[0].grid(True, linestyle="--", alpha=0.5)

# Average sales comparison
axes[1].bar(["Product A", "Product B"], [avg_a, avg_b], color=["steelblue", "darkorange"])
axes[1].set_title("Average monthly sales")
axes[1].set_ylabel("Average sales")
axes[1].grid(axis="y", linestyle="--", alpha=0.5)

# Total sales share
axes[2].pie(product_total, labels=product_names, autopct="%1.1f%%", startangle=90,
            colors=["steelblue", "darkorange"])
axes[2].set_title("Total sales share")

plt.tight_layout()
plt.show()
```
