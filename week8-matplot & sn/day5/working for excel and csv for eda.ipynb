`````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.barplot(data=df, x="name", y="salary", hue="age")   
sns.histplot(data=df, x="name",
             y="salary", hue="age", bins=10)
sns.set(style="darkgrid")

plt.show()
`````

`````python
correlation = df.corr(numeric_only=True)
`````

`````python
sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.savefig("C:\\Users\\olaaz\\Documents\\AI Roadmap\\week8\\day5\\correlation_heatmap.png", dpi=300, bbox_inches='tight')
plt.close()
`````
