```python
import seaborn as sns
import matplotlib.pyplot as plt

salaries = [
    3000, 3200, 3500, 3600,
    3700, 3800, 4000, 4200,
    4500, 4800, 5000, 15000
]
sns.histplot(salaries, bins=6)
sns.boxplot(x=salaries, color='lightblue')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cities =[
    "dubai",
    "Abu Dhabi",
    "Sharjah",
    "Ajman"
]

sales = [400, 320, 270, 167]
sns.barplot(x=cities, y=sales)
plt.title("Sales by Cities")
plt.xlabel("Cities")
plt.ylabel("Sales")
plt.show()
```

```python

import pandas as pd
df = pd.DataFrame({
    "hours": [1, 2, 4, 5, 6, 7, 8, 9],
    "score": [50, 55, 60, 70, 78, 85, 90, 99]

})
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.scatterplot(
    data=df,
    x="hours",
    y="score"
)
plt.title("Hours vs Score")
sns.barplot(
    x=df["hours"],
    y=df["score"]
)
plt.show()
```

```python
df = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5, 6],
    "score": [50, 55, 60, 70, 78, 85],
    "level": [
        "Beginner",
        "Beginner",
        "Beginner",
        "Advanced",
        "Advanced",
        "Advanced"
    ]
})
sns.scatterplot(
    data=df,
    x="hours",
    y="score",
    hue="level"
)
sns.barplot(
    x=df["hours"],
    y=df["score"],
    hue=df["level"]
)

plt.show()
```
