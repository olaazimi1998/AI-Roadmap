```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 5, 6, 7, 8, 15, 20]
sales = [10, 20, 30, 40, 50, 60, 30, 50, 79]
plt.plot(days, sales)
plt.xlabel("days")
plt.ylabel("sales")
plt.show()
```

```python
plt.title("Daily sales")
plt.plot(days, sales)
plt.xlabel("Day")
plt.ylabel("sales")
plt.scatter(days, sales)
plt.bar(days, sales)
plt.legend()
plt.show()
```

```python
ages = [18, 19, 20, 21, 22, 23, 24, 26, 27, 29,]
plt.hist(ages)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]

product_a = [10, 15, 20, 25, 30]
product_b = [8, 18, 16, 28, 35]

plt.plot(months, product_a, label="Product A")
plt.plot(months, product_b, label="Product B")

plt.title("Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.bar(months, product_a)
plt.legend()

plt.show()
```
