Word2Vec یکی از روش‌های معروف برای ساخت Word Embedding است.

ایده اصلی:

Words that appear in similar contexts tend to have similar meanings.

یعنی:

کلماتی که در محیط‌های مشابه استفاده می‌شوند، معمولاً معنای مشابهی دارند.

 Two Word2Vec approaches
Word2Vec معمولاً با دو روش معروف معرفی می‌شود:

CBOW
Continuous Bag of Words

از کلمات اطراف برای پیش‌بینی کلمه‌ی وسط استفاده می‌کند.

مثلاً:

I drink ___ every morning.
مدل تلاش می‌کند:

coffee
را پیش‌بینی کند.

```python
from gensim.models import Word2Vec

sentences = [
    ["i", "love", "python"],
    ["i", "love", "machine", "learning"],
    ["python", "is", "powerful"],
    ["machine", "learning", "is", "interesting"],
    ["i", "like", "python"],
]
```

```python
model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1)
```

```python

print("Vocabulary:")
print(model.wv.index_to_key)

print("\nVector for python:")
print(model.wv["python"])
```

```python
from gensim.models import Word2Vec


sentences = [
    ["i", "love", "python"],
    ["i", "love", "machine", "learning"],
    ["python", "is", "powerful"],
    ["machine", "learning", "is", "interesting"],
    ["i", "like", "python"],
]

model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1
)

print("Vocabulary:")
print(model.wv.index_to_key)

print("\nVector for python:")
print(model.wv["python"])

print("\nSimilar words to python:")
print(model.wv.most_similar("python"))
```

| Word2Vec                 | GloVe                                      |
| ------------------------ | ------------------------------------------ |
| Word embedding method    | Word embedding method                      |
| Uses local context       | Uses global word co-occurrence information |
| Learns from nearby words | Uses statistics from word co-occurrences   |
| Commonly used            | Commonly used                              |

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np


python = np.array([[0.9, 0.8, 0.7]])

java = np.array([[0.85, 0.75, 0.65]])

banana = np.array([[0.1, -0.5, 0.2]])


python_java = cosine_similarity(python, java)

python_banana = cosine_similarity(python, banana)


print("Python vs Java:")
print(python_java)

print("\nPython vs Banana:")
print(python_banana)
```
