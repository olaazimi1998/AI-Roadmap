```python
import torch
import torch.nn as nn
```

```python
X = torch.tensor([
    [[1.0], [2.0], [3.0], [4.0]],
    [[2.0], [3.0], [4.0], [5.0]],
    [[3.0], [4.0], [5.0], [6.0]],
])

y = torch.tensor([
    [5.0],
    [6.0],
    [7.0]
])
```

```python
print(X.shape)
```

3 = number of samples
4 = sequence length
1 = number of features

```python

class LSTMModel(nn.Module):
    def __init__(self):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=32,
            num_layers=1,
            batch_first=True
        )

        self.fc = nn.Linear(32, 1)

    def forward(self, x):
        output, (hidden, cell) = self.lstm(x)
        prediction = self.fc(output[:, -1, :])
        return prediction

model = LSTMModel()

prediction = model(X)

print("Prediction:")
print(prediction)

print("Prediction shape:")
print(prediction.shape)
        
```
