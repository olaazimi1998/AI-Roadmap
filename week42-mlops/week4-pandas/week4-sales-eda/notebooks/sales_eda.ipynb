```python
df["sales"] = pd.to_numeric(df["sales"], errors="coerce")
```

```python
df["sales"].isnull().sum()
```

```python
df["region"] = df["region"].str.strip().str.title()
df["product"] = df["product"].str.strip().str.title()
```

```python
df.describe()
```

```python
df["region"] = df["region"].replace({"Dubaii": "Dubai"})
```

```python
df[df["sales"] < 0]
```

```python
df[df["quantity"] < 0]
```

```python
df["sales"] = df["sales"].fillna(df["sales"].median())
df["region"] = df["region"].fillna("unknown")
```

```python

df["order_id"] = df["order_id"].fillna("unknown")
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
sales_by_region = (df.groupby("region")["sales"].sum().sort_values(ascending=False))
sales_by_region
```

```python
sales_by_product = (df.groupby("product")["sales"].sum().sort_values(ascending=False))
```

```python
sales_by_product
```

```python
df["sales"].mean()
```

```python
df.groupby(["region", "product"])["sales"].sum()
```

```python
pivot = pd.pivot_table(df, values="sales", index="region", columns="product", aggfunc="sum", fill_value=0)
```

```python
top_region = sales_by_region.idxmax()
top_region_sales = sales_by_region.max()

print("Top region:", top_region)
print("Total sales:", top_region_sales)
```

```python
top_product = sales_by_product.idxmax()
top_product_sales = sales_by_product.max()

print("Top product:", top_product)
print("Total sales:", top_product_sales)
```
