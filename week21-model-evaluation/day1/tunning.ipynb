Model Evaluation & Hyperparameter Tuning
According to your roadmap, Week 16 is about evaluating ML models correctly and tuning their hyperparameters. The week covers accuracy, precision, recall, F1, ROC-AUC, RMSE, k-fold cross-validation, GridSearchCV, RandomizedSearchCV, and bias-variance tradeoff. 

|                        | Actual Positive | Actual Negative |
| ---------------------- | --------------: | --------------: |
| **Predicted Positive** |              TP |              FP |
| **Predicted Negative** |              FN |              TN |


Accuracy = Correct Predictions / All Predictions

Accuracy = (TP + TN) / (TP + TN + FP + FN)

recision asks:

When my model says Positive, how often is it correct?

Formula:

Precision = TP / (TP + FP)
Example:

TP = 80
FP = 10

ecall asks:

Of all the actual positive cases, how many did my model find?

Formula:

Recall = TP / (TP + FN)
Example:

TP = 80
FN = 20

TP = 80
FN = 20
Then:

Recall = 80 / (80 + 20)
       = 0.80
So:

Recall = 80%
Easy way to remember
Recall → "Did I find most of the real positives?"



etimes precision and recall are both important.

F1 combines them.

F1 = 2 × (Precision × Recall) / (Precision + Recall)
For example:

Precision = 0.889
Recall = 0.800
Then:

F1 ≈ 0.842
So:

F1 ≈ 84.2%

Accuracy
    ↓
How many predictions were correct?

Precision
    ↓
When I predict Positive, am I correct?

Recall
    ↓
Did I find the actual Positives?

F1
    ↓
Balance between Precision and Recall

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
```

```python
y_true = [1, 1, 0, 0, 1, 0, 1, 1, 0, 0]

y_pred = [1, 1, 0, 1, 1, 0, 0, 1, 0, 0]
```

```python
accuracy = accuracy_score(y_true, y_pred)

print("Accuracy:", accuracy)
```

```python
precision = precision_score(y_true, y_pred)

print("Precision:", precision)
```

```python
recall = recall_score(y_true, y_pred)

print("Recall:", recall)

f1 = f1_score(y_true, y_pred)

print("F1:", f1)
```

```python
cm = confusion_matrix(y_true, y_pred)

print(cm)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Real values
y_true = [1, 1, 0, 0, 1, 0, 1, 1, 0, 0]

# Model predictions
y_pred = [1, 1, 0, 1, 1, 0, 0, 1, 0, 0]

# Metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

# Confusion matrix
cm = confusion_matrix(y_true, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)

print("\nConfusion Matrix:")
print(cm)
```
