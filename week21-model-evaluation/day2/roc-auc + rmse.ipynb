```python
from sklearn.metrics import roc_auc_score

y_true = [0, 0, 1, 1, 1]

y_prob = [0.1, 0.2, 0.7, 0.8, 0.9]

auc = roc_auc_score(y_true, y_prob)

print("ROC-AUC:", auc)
```

AUC close to 1.0 → very good separation
AUC around 0.5  → random guessing
AUC below 0.5   → very poor separation

 What is ROC-AUC?
ROC-AUC helps us understand how well a classification model separates two classes.

RMSE is mainly used for regression.

For example, imagine we want to predict salary:

Real salary       Predicted salary

5000              4800
7000              7200
4000              3900
6000              6500
RMSE tells us approximately how far our predictions are from the real values.

```python
from sklearn.metrics import root_mean_squared_error

y_true = [5000, 7000, 4000, 6000]

y_pred = [4800, 7200, 3900, 6500]

rmse = root_mean_squared_error(
    y_true,
    y_pred
)

print("RMSE:", rmse)
```
