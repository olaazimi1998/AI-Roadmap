```python
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression


# ==================================================
# PART A — HOUSE PRICE
# ==================================================

X_price = np.array([
    [800],
    [1000],
    [1200],
    [1500],
    [1800]
])

y_price = np.array([
    200000,
    250000,
    300000,
    375000,
    450000
])


linear_model = LinearRegression()

linear_model.fit(X_price, y_price)


house_size = 1300

price_prediction = linear_model.predict(
    [[house_size]]
)


print("===== HOUSE PRICE =====")

print("House size:", house_size)

print("Predicted price:", price_prediction[0])


# ==================================================
# PART B — SURVIVAL
# ==================================================

X_survival = np.array([
    [10],
    [20],
    [30],
    [40],
    [50],
    [60]
])

y_survival = np.array([
    0,
    0,
    0,
    1,
    1,
    1
])


logistic_model = LogisticRegression()

logistic_model.fit(X_survival, y_survival)


age = 45


prediction = logistic_model.predict(
    [[age]]
)


probability = logistic_model.predict_proba(
    [[age]]
)


print("\n===== SURVIVAL =====")

print("Age:", age)

print("Prediction:", prediction[0])

print("Probability:", probability[0])
```

```python
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression


# ==================================================
# LINEAR REGRESSION
# ==================================================

X_price = np.array([
    [800],
    [1000],
    [1200],
    [1500],
    [1800]
])

y_price = np.array([
    200000,
    250000,
    300000,
    375000,
    450000
])


model = LinearRegression()

model.fit(X_price, y_price)


print("===== LINEAR REGRESSION =====")

print("Weight:", model.coef_[0])

print("Bias:", model.intercept_)

print(
    "Prediction for 1300:",
    model.predict([[1300]])[0]
)


# ==================================================
# LOGISTIC REGRESSION
# ==================================================

X_survival = np.array([
    [10],
    [20],
    [30],
    [40],
    [50],
    [60]
])

y_survival = np.array([
    0,
    0,
    0,
    1,
    1,
    1
])


model = LogisticRegression()

model.fit(X_survival, y_survival)


print("\n===== LOGISTIC REGRESSION =====")

print("Weight:", model.coef_[0][0])

print("Bias:", model.intercept_[0])

print(
    "Prediction for age 45:",
    model.predict([[45]])[0]
)

print(
    "Probability for age 45:",
    model.predict_proba([[45]])[0]
)
```
