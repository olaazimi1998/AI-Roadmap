### Linear regression from scrach


```python
import numpy as np

# Training data
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)

# Parameters
w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000

n = len(X)

for epoch in range(epochs):

    # Prediction
    y_pred = w * X + b

    # Error
    error = y_pred - y

    # Gradients
    dw = (2 / n) * np.sum(X * error)
    db = (2 / n) * np.sum(error)

    # Update
    w -= learning_rate * dw
    b -= learning_rate * db

print("Weight:", w)
print("Bias:", b)

# Prediction
x_new = 6

prediction = w * x_new + b

print("Prediction:", prediction)

```

```python
import numpy as np


def sigmoid(z):
    return 1 / (1 + np.exp(-z))


X = np.array([1, 2, 3, 4, 5, 6], dtype=float)
y = np.array([0, 0, 0, 1, 1, 1], dtype=float)

w = 0.0
b = 0.0

learning_rate = 0.1
epochs = 1000

n = len(X)

for epoch in range(epochs):

    # Linear combination
    z = w * X + b

    # Probability
    y_pred = sigmoid(z)

    # Gradients
    dw = (1 / n) * np.sum(X * (y_pred - y))
    db = (1 / n) * np.sum(y_pred - y)

    # Update
    w -= learning_rate * dw
    b -= learning_rate * db


print("Weight:", w)
print("Bias:", b)


# Test
x_new = 5

probability = sigmoid(w * x_new + b)

prediction = 1 if probability >= 0.5 else 0

print("Probability:", probability)
print("Prediction:", prediction)
```

```python
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression


# --------------------
# Linear Regression
# --------------------

X_price = np.array([
    [800],
    [1000],
    [1200],
    [1500],
    [1800]
])

y_price = np.array([
    200000,
    250000,
    300000,
    375000,
    450000
])

linear_model = LinearRegression()

linear_model.fit(X_price, y_price)

price_prediction = linear_model.predict([[1300]])

print("Predicted house price:", price_prediction)


# --------------------
# Logistic Regression
# --------------------

X_survival = np.array([
    [10],
    [20],
    [30],
    [40],
    [50],
    [60]
])

y_survival = np.array([
    0,
    0,
    0,
    1,
    1,
    1
])

logistic_model = LogisticRegression()

logistic_model.fit(X_survival, y_survival)

prediction = logistic_model.predict([[45]])

probability = logistic_model.predict_proba([[45]])

print("Survival prediction:", prediction)
print("Probability:", probability)
```
