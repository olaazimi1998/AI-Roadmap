Today we will actually code Gradient Descent using NumPy.

w = w - learning_rate × dw

 📚 What Is Learning Rate?
We need to decide how big each step should be.

That's the:

```python
import numpy as np
```

```python
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])
w = 0
b = 0

learning_rate = 0.01
```

```python
epochs = 5
for i in range(epochs):

    predictions = w * X + b

    error = y - predictions

    dw = (-2 / len(X)) * np.sum(X * error)

    db = (-2 / len(X)) * np.sum(error)

    w = w - learning_rate * dw

    b = b - learning_rate * db

    print("Final w:", w)
    print("Final b:", b)
    
    print("Predictions:", w * X + b)
```

```python
import numpy as np


X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

w = 0
b = 0

learning_rate = 0.01
epochs = 1000


for i in range(epochs):

    predictions = w * X + b

    error = y - predictions

    mse = np.mean(error ** 2)

    dw = (-2 / len(X)) * np.sum(X * error)
    db = (-2 / len(X)) * np.sum(error)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    if i % 100 == 0:
        print("Epoch:", i, "MSE:", mse)


print()
print("Final w:", w)
print("Final b:", b)
print("Final predictions:", w * X + b)
```
