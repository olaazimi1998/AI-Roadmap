2. ReLU
ReLU is very simple:

ReLU(x) = max(0, x)
That means:

x = -5  → 0
x = -2  → 0
x =  0  → 0
x =  3  → 3
x =  8  → 8
So:

negative → 0
positive → same number

 Sigmoid
Sigmoid converts a number into a value between:

0 and 1
Conceptually:

large negative → close to 0

0 → 0.5

large positive → close to 1
This makes sigmoid useful when we want something that behaves like a probability for a binary output

```python
import math


def sigmoid(x):
    return 1 / (1 + math.exp(-x))


print(sigmoid(-5))
print(sigmoid(0))
print(sigmoid(5))
```

Softmax
Softmax is commonly used when we have multiple classes.

Imagine our model predicts:

Cat
Dog
Bird
The network might produce:

Cat   → 2.5
Dog   → 1.2
Bird  → 0.3

Softmax converts these into values that behave like probabilities and sum to approximately:

1.0
For example:

Cat   → 0.70
Dog   → 0.20
Bird  → 0.10
