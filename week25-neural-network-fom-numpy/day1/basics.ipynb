```python
import numpy as np

x = np.array([[2, 3]])

weights = np.array([
    [0.5],
    [0.2]
])

bias = 0.1

z = np.dot(x, weights) + bias

print("Z:", z)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))


output = sigmoid(z)

print("Output:", output)
```

```python
class NeuralNetwork:

    def __init__(self):
        self.weights_input_hidden = np.random.randn(2, 4)
        self.bias_hidden = np.zeros((1, 4))

        self.weights_hidden_output = np.random.randn(4, 1)
        self.bias_output = np.zeros((1, 1))
```

Input → Hidden
We have:

2 inputs
4 hidden neurons
Therefore:

self.weights_input_hidden

Hidden → Output
We have:

4 hidden neurons
1 output neuron

```python
def sigmoid(self, x):
    return 1 / (1 + np.exp(-x))
```

```python
def forward(self, X):

    hidden_input = np.dot(X, self.weights_input_hidden)
    hidden_input = hidden_input + self.bias_hidden

    hidden_output = self.sigmoid(hidden_input)

    final_input = np.dot(hidden_output, self.weights_hidden_output)
    final_input = final_input + self.bias_output

    final_output = self.sigmoid(final_input)

    return final_output
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

network = NeuralNetwork()

class NeuralNetwork:

    def __init__(self):
        self.weights_input_hidden = np.random.randn(2, 4)
        self.bias_hidden = np.zeros((1, 4))

        self.weights_hidden_output = np.random.randn(4, 1)
        self.bias_output = np.zeros((1, 1))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def forward(self, X):
        hidden_input = np.dot(X, self.weights_input_hidden) + self.bias_hidden
        hidden_output = self.sigmoid(hidden_input)

        final_input = np.dot(hidden_output, self.weights_hidden_output) + self.bias_output
        final_output = self.sigmoid(final_input)

        return final_output

predictions = network.forward(X)
print(predictions)
```
